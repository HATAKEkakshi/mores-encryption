# mores-encryption Integration Guide

This notebook demonstrates how to use the `mores-encryption` library for secure data handling in your application.

```python
import os
from mores_encryption.encryption import encryption_service
import json
```

## 1. Basic Encryption and Decryption

Use `encrypt` to secure sensitive strings and `decrypt` to recover them.

```python
sensitive_data = "My Secret Credit Card Number"

# Encrypt
encrypted_text = encryption_service.encrypt(sensitive_data)
print(f"Original: {sensitive_data}")
print(f"Encrypted: {encrypted_text}")

# Decrypt
decrypted_text = encryption_service.decrypt(encrypted_text)
print(f"Decrypted: {decrypted_text}")

assert sensitive_data == decrypted_text
```

## 2. Deterministic Hashing for Search

Standard encryption produces different outputs for the same input (due to IVs). 
To search for encrypted data (e.g., "Find user with email X"), use `hash` with a constant salt.

```python
email = "user@example.com"
salt = "static_application_salt_value_123" # Must be constant for the app

email_hash_1 = encryption_service.hash(email, salt)
print(f"Hash 1: {email_hash_1}")

# Hashing the same email with the same salt produces the SAME output
email_hash_2 = encryption_service.hash(email, salt)
print(f"Hash 2: {email_hash_2}")

assert email_hash_1 == email_hash_2
```

## 3. Handling Different Salts
Changing the salt changes the hash completely. This prevents rainbow table attacks.

```python
different_salt = "other_salt_value_999"
email_hash_3 = encryption_service.hash(email, different_salt)
print(f"Hash with different salt: {email_hash_3}")

assert email_hash_1 != email_hash_3
```

# 4. Handling JSON DATA

Mores Encryption can be used to encrypt and decrypt JSON data as well.  Not directly you need to convert the JSON data to string and then encrypt it. Mores Encryption provides a helper function to handle this.

```python
sensitive_data = {
    "My Secret Credit Card Number": "1234-5678-9012-3456",
    "My Secret Password": "password123"
}

# Encrypt JSON directly
encrypted_text = encryption_service.encrypt_json(sensitive_data)
print(f"Original: {sensitive_data}")
print(f"Encrypted: {encrypted_text}")

# Decrypt JSON directly
decrypted_data = encryption_service.decrypt_json(encrypted_text)
print(f"Decrypted Data: {decrypted_data}")

assert sensitive_data == decrypted_data
print(f"Type of encrypted text: {type(encrypted_text)} and type of decrypted data: {type(decrypted_data)}")
```
