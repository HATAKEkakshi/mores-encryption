# Determinstic Hash
Deterministic Hash creates a secure hash of the data that is consistent and repeatable. Unlike standard encryption (which produces different outputs for the same input), deterministic hashing ensures that the same input always generates the exact same hash. This allows you to securely search for sensitive data (like emails or IDs) without verifying the plaintext.

```python
email="test.example@xyz.in"
hashed_email=ep.encryption_service.hash(data=email,salt="589VKnNGNCdGqYKwe6bDpQ",iterations=300_000)
hashed_email
```

## Testing the email hashed before and Now is same
